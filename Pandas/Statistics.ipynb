```python
import pandas as pd

data = {
    "Name": ["Nitu", "Rahul", "Priya", "Amit", "Sneha", "Nitu"],
    "Age": [24, 28, 26, 31, 29, 24],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [35000, 45000, 40000, 60000, 48000, 35000],
    "City": ["Pune", "Mumbai", "Pune", "Delhi", "Mumbai", "Pune"],
    "Joining_Date": [
        "2024-01-15",
        "2022-06-10",
        "2023-03-20",
        "2019-08-12",
        "2021-11-05",
        "2024-01-15"
    ]
}

df = pd.DataFrame(data)
df
```

```python
#Q1. Find the average salary.
df["Salary"].mean()
```

```python
#Q2. Find the median salary.
df["Salary"].median()
```

```python
#Q3. Find the minimum salary.
df["Salary"].min()
```

```python
#Q4. Find the maximum salary.
df["Salary"].max()
```

```python
#Q5. Find the average age.
df["Age"].mean()
```

```python
#Q6. Find the standard deviation of salary.
df["Salary"].std()
```

```python
#Q7. Find the variance of salary.
df["Salary"].var()
```

```python
#Q8. Use describe() on the DataFrame.
df.describe()
```

```python
#Q9. Find the average salary department-wise.
df.groupby("Department")["Salary"].mean()
```

```python
#Q10. Find the highest salary in each department.
df.groupby("Department")["Salary"].max()
```
