```python
#Melt Function
```

```python
import pandas as pd

var = pd.DataFrame({
    "Days":[1,2,3,4,5],
    "Eng":[12,13,14,15,15],
    "Math":[21,3,2,43,21]
})
var
```

```python
pd.melt(var)
```

```python
pd.melt(var,id_vars=["Math"])
```

```python
pd.melt(var,id_vars=["Days"])
```

```python
pd.melt(var,id_vars=["Days"],var_name="Nitu")
```

```python
pd.melt(var,id_vars=["Days"],var_name="IIII",value_name="OOOO")
```

```python
#Pivot Table
```

```python
cr = pd.DataFrame({
    "Days":[1,2,3,4,5],
    "stu":['a','b','c','d','e'],
    "Eng":[12,13,14,15,15],
    "Math":[21,3,2,43,21]
})
cr
```

```python
cr.pivot(index="Days",columns="stu")
```

```python
sr = pd.DataFrame({
    "Days":[1,2,3,4,5],
    "stu":['a','a','c','c','e'],
    "Eng":[12,13,14,15,15],
    "Math":[21,3,2,43,21]
})
sr
```

```python
sr.pivot(index="Days",columns="stu")
```

```python
sr.pivot(index="Days",columns="stu",values="Eng")
```

```python
sr.pivot(index="Days",columns="stu")
```

```python
mr = pd.DataFrame({
    "Days":[1,1,2,1,2],
    "stu":['a','a','c','c','e'],
    "Eng":[12,13,14,15,15],
    "Math":[21,3,2,43,21]
})
mr
```

```python
mr.pivot_table(index="stu",columns="Days",aggfunc="mean")
```

```python
mr.pivot_table(index="stu",columns="Days",aggfunc="sum")
```

```python
mr.pivot_table(index="stu",columns="Days",aggfunc="median",margins=True)
```
