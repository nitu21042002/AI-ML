```python
#Merge
```

```python
import pandas as pd

var = pd.DataFrame({"A":[1,2,3,4],"B":[4,7,5,6]})
val = pd.DataFrame({"A":[1,2,3,4],"C":[11,22,33,44]})

pd.merge(var,val,on="A")
```

```python
var = pd.DataFrame({"A":[1,2,3,4],"B":[4,7,5,6]})
val = pd.DataFrame({"A":[1,2,3,5],"C":[11,22,33,44]})

pd.merge(var,val,on="A")
```

```python
pd.merge(var,val,how="left")
```

```python
pd.merge(var,val,how="right")
```

```python
pd.merge(var,val,how="cross")
```

```python
pd.merge(var,val,how="inner")
```

```python
pd.merge(var,val,how="outer")
```

```python
pd.merge(var,val,how="left_anti")
```

```python
pd.merge(var,val,how="right_anti")
```

```python
pd.merge(var,val,left_index=True,right_index=True)
```

```python
pd.merge(var,val,left_index=True,right_index=True,suffixes=("name","python"))
```

```python
#Concat
```

```python
sr = pd.Series([1,2,3,4])
cr = pd.Series([11,22,33,44])
pd.concat([sr,cr])
```

```python
va1 = pd.DataFrame({"A":[1,2,3,4],"B":[4,7,5,6]})
va2 = pd.DataFrame({"A":[1,2,3,4],"B":[11,22,33,44]})
pd.concat([va1,va2])
```

```python
va1 = pd.DataFrame({"A":[1,2,3,4],"B":[4,7,5,6]})
va2 = pd.DataFrame({"A":[1,2,3,4],"C":[11,22,33,44]})
pd.concat([va1,va2])
```

```python
pd.concat([va1, va2],axis=1)
```

```python
pd.concat([va1, va2],axis=1,join="outer")
```

```python
pd.concat([va1, va2],axis=1,keys=["va1","va2"])
```
