```python
import pandas as pd

data = {
    "Name": ["Nitu", "Rahul", "Priya", "Amit", "Sneha", "Nitu"],
    "Age": [24, 28, 26, 31, 29, 24],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [35000, 45000, 40000, 60000, 48000, 35000],
    "City": ["Pune", "Mumbai", "Pune", "Delhi", "Mumbai", "Pune"],
    "Joining_Date": [
        "2024-01-15",
        "2022-06-10",
        "2023-03-20",
        "2019-08-12",
        "2021-11-05",
        "2024-01-15"
    ]
}

df = pd.DataFrame(data)
df
```

```python
#Q1. Convert all City names to uppercase.
df["City"] = df["City"].str.upper()
df
```

```python
#Q2. Convert all employees names to lowercase.
df["Name"] = df["Name"].str.lower()
df
```

```python
#Q3. Find employees whose name contains "i".
df[df["Name"].str.contains("i", case=False)]
```

```python
#Q4. Find employees whose city contains "Pune".
df[df["City"].str.contains("Pune", case=False)]
df
```

```python
#Q5. Find the length of each employee's name.
df["Name_Length"] = df["Name"].str.len()
df
```

```python
#Q6. Replace "Pune" with "Mumbai".
df["City"] = df["City"].str.replace("Pune", "Mumbai")
df
```
