```python
#Group By
```

```python
import pandas as pd

var = pd.DataFrame({"name":["a","a","b",'f','f','c','b','b','d',],
                     "sub1":[12,13,12,13,14,15,16,14,54],
                     "sub2":[21,32,43,23,34,54,45,65,56]
                    })
var

```

```python
new_var = var.groupby("name")
new_var
```

```python
for x,y in new_var:
    print(x)
    print(y)
```

```python
new_var.get_group("a")
```

```python
new_var.get_group("f")
```

```python
new_var.min()
```

```python
new_var.max()
```
