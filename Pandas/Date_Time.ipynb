```python
import pandas as pd

data = {
    "Name": ["Nitu", "Rahul", "Priya", "Amit", "Sneha", "Nitu"],
    "Age": [24, 28, 26, 31, 29, 24],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [35000, 45000, 40000, 60000, 48000, 35000],
    "City": ["Pune", "Mumbai", "Pune", "Delhi", "Mumbai", "Pune"],
    "Joining_Date": [
        "2024-01-15",
        "2022-06-10",
        "2023-03-20",
        "2019-08-12",
        "2021-11-05",
        "2024-01-15"
    ]
}

df = pd.DataFrame(data)
df
```

```python
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])
df
```

```python
#Q1. Extract the joining year.
df["Joining_Year"] = df["Joining_Date"].dt.year
df
```

```python
#Q2. Extract the joining month.
df["Joining_Month"] = df["Joining_Date"].dt.month
df
```

```python
#Q3. Extract the joining day.
df["Joining_Day"] = df["Joining_Date"].dt.day
df
```

```python
#Q4. Find employees who joined after 2022.
df[df["Joining_Date"].dt.year > 2022]
```

```python
#Q5. Find employees who joined in 2024.
df[df["Joining_Date"].dt.year == 2024]
```

```python
#Q6. Calculate how many years each employee has been working.
df["Working_Years"] = (
    (pd.Timestamp.today() - df["Joining_Date"]).dt.days / 365.25
).astype(int)
df
```
