```python
import pandas as pd

data = {
    "Name": ["Nitu", "Rahul", "Priya", "Amit", "Sneha", "Nitu"],
    "Age": [24, 28, 26, 31, 29, 24],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [35000, 45000, 40000, 60000, 48000, 35000],
    "City": ["Pune", "Mumbai", "Pune", "Delhi", "Mumbai", "Pune"],
    "Joining_Date": [
        "2024-01-15",
        "2022-06-10",
        "2023-03-20",
        "2019-08-12",
        "2021-11-05",
        "2024-01-15"
    ]
}

df = pd.DataFrame(data)
df
```

```python
#Q1. Find duplicate rows.
df.duplicated()
```

```python
#Q2. Count the number of duplicate rows.
df.duplicated().sum()
```

```python
#Q3. Remove duplicate rows.
df.drop_duplicates()
```

```python
#Q4. Remove duplicates based only on Name.
df.drop_duplicates(subset=["Name"])
```

```python
#Q5. Keep the last occurrence instead of the first.
df.drop_duplicates(keep="last")
```
