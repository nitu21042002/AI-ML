```python
import pandas as pd

data = {
    "Name": ["Nitu", "Rahul", "Priya", "Amit", "Sneha", "Nitu"],
    "Age": [24, 28, 26, 31, 29, 24],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [35000, 45000, 40000, 60000, 48000, 35000],
    "City": ["Pune", "Mumbai", "Pune", "Delhi", "Mumbai", "Pune"],
    "Joining_Date": [
        "2024-01-15",
        "2022-06-10",
        "2023-03-20",
        "2019-08-12",
        "2021-11-05",
        "2024-01-15"
    ]
}

df = pd.DataFrame(data)
df
```

```python
#Q1. Check the data types of all columns.
df.dtypes
```

```python
#Q2. Convert Age to float.
df["Age"] = df["Age"].astype(float)
df
```

```python
#Q3. Convert Salary to float.
df["Salary"] = df["Salary"].astype(float)
df
```

```python
#Q4. Convert Joining_Date from object/string to datetime.
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])
df
```

```python
#Q5. Check the data types again.
df.dtypes
```
