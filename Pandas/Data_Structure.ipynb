```python
#Series
```

```python
import pandas as pd
x=[1,4,5,8,2,5]
var = pd.Series(x)
print(var)
```

```python
var = pd.Series(x,index=['a','b','c','d','e','f'],dtype='float',name="python")
print(var)
print(type(var))
print(var['d'])
```

```python
dic = {
    "Name":["nitu","ritu","rinaa"],
    "age":[21,22,23],
    "marks":[98,99,87]
}
print(pd.Series(dic))
```

```python
s=pd.Series(12,index=[0,1,2,3,4,5])
print(s,type(s))
```

```python
s=pd.Series(12,index=[0,1,2,3,4,5])
a=pd.Series(12,index=[3,2,4])
print(s+a,type(s),type(a))
```
