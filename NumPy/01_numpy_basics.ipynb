```python
# Q10. Find the dimension, shape, size, item size, and total bytes of an array.

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Dimension:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Item size:", arr.itemsize)
print("Total bytes:", arr.nbytes)
```
