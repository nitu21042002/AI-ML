```python
va1 = pd.DataFrame({"A":[1,2,3,4],"B":[7,5,6,7]},index=['a','b','c','d'])
va2 = pd.DataFrame({"C":[21,22],"D":[11,22]},index=['a','b'])

va1.join(va2)
```

```python
va1.join(va2,how="cross")
```

```python
va1.join(va2,how="left")
```

```python
va1.join(va2,how="right")
```

```python
va1.join(va2,how="inner")
```

```python
va2.join(va1,how="outer",lsuffix="BR",rsuffix="BR")
```

```python
va1.rename(columns={"B": "B_B"}, inplace=True)
va1
```
