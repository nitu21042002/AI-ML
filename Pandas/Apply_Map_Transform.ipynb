```python
import pandas as pd

data = {
    "Name": ["Nitu", "Rahul", "Priya", "Amit", "Sneha", "Nitu"],
    "Age": [24, 28, 26, 31, 29, 24],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [35000, 45000, 40000, 60000, 48000, 35000],
    "City": ["Pune", "Mumbai", "Pune", "Delhi", "Mumbai", "Pune"],
    "Joining_Date": [
        "2024-01-15",
        "2022-06-10",
        "2023-03-20",
        "2019-08-12",
        "2021-11-05",
        "2024-01-15"
    ]
}

df = pd.DataFrame(data)
df
```

```python
#Q1. Use apply() to increase every salary by 10%.
df["Salary"] = df["Salary"].apply(lambda x: x * 1.10)
df
```

```python
#Q2. Use apply() to create a new column: Salary > 50000 → "High" Otherwise → "Low"
df["Salary_Level"] = df["Salary"].apply(
    lambda x: "High" if x > 50000 else "Low"
)
df
```

```python
#Q3. Use map() to change: IT → Information Technology HR → Human Resources Finance → Financial Services
df["Department"] = df["Department"].map({
    "IT": "Information Technology",
    "HR": "Human Resources",
    "Finance": "Financial Services"
})
df
```

```python
#Q4. Use transform() to calculate the average salary.
df["Average_Salary"] = df.groupby(lambda x: True)["Salary"].transform("mean")
df
```
