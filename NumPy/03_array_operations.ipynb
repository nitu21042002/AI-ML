```python
# Q21. Add two NumPy arrays element by element.

import numpy as np

arr1 = np.array([10, 20, 30])
arr2 = np.array([1, 2, 3])

print(arr1 + arr2)
```

```python
# Q22. Subtract one array from another.

import numpy as np

arr1 = np.array([10, 20, 30])
arr2 = np.array([1, 2, 3])

print(arr1 - arr2)
```

```python
# Q23. Multiply and divide two arrays.

import numpy as np

arr1 = np.array([10, 20, 30])
arr2 = np.array([2, 4, 5])

print("Multiplication:", arr1 * arr2)
print("Division:", arr1 / arr2)
```

```python
# Q24. Find the sum of all elements in an array.

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(np.sum(arr))
```

```python
# Q25. Find the minimum and maximum values.

import numpy as np

arr = np.array([15, 5, 35, 10, 25])

print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
```

```python
# Q26. Calculate the mean, median, and standard deviation.

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
```

```python
# Q27. Calculate the cumulative sum of an array.

import numpy as np

arr = np.array([10, 20, 30, 40])

print(np.cumsum(arr))
```

```python
# Q28. Sort an array in ascending and descending order.

import numpy as np

arr = np.array([40, 10, 30, 20, 50])

print("Ascending:", np.sort(arr))
print("Descending:", np.sort(arr)[::-1])
```

```python
# Q29. Find the index of the maximum element.

import numpy as np

arr = np.array([10, 50, 20, 80, 30])

print(np.argmax(arr))
```

```python
# Q30. Count how many elements are greater than 25.

import numpy as np

arr = np.array([10, 30, 20, 40, 50, 15])

count = np.sum(arr > 25)

print(count)
```
