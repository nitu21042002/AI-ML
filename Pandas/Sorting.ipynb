```python
import pandas as pd

data = {
    "Name": ["Nitu", "Rahul", "Priya", "Amit", "Sneha", "Nitu"],
    "Age": [24, 28, 26, 31, 29, 24],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [35000, 45000, 40000, 60000, 48000, 35000],
    "City": ["Pune", "Mumbai", "Pune", "Delhi", "Mumbai", "Pune"],
    "Joining_Date": [
        "2024-01-15",
        "2022-06-10",
        "2023-03-20",
        "2019-08-12",
        "2021-11-05",
        "2024-01-15"
    ]
}

df = pd.DataFrame(data)
df
```

```python
#Q1. Sort employees by Salary from lowest to highest.
df.sort_values("Salary")
```

```python
#Q2. Sort employees by Salary from highest to lowest.
df.sort_values("Salary", ascending=False)
```

```python
#Q3. Sort employees by Age.
df.sort_values("Age")
```

```python
#Q4. Sort employees by Department and then Salary.
df.sort_values(["Department", "Salary"])
```

```python
#Q5. Sort the DataFrame by index.
df.sort_index()
```
