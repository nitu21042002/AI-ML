```python
# Q51. Calculate the average, highest, and lowest student marks.

import numpy as np

marks = np.array([75, 85, 60, 90, 70])

print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))
```

```python
# Q52. Find students who scored more than 75 marks.

import numpy as np

marks = np.array([65, 80, 72, 95, 60, 85])

result = marks[marks > 75]

print(result)
```

```python
# Q53. Normalize an array using min-max normalization.

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

normalized = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

print(normalized)
```

```python
# Q54. Calculate the percentage of missing values in an array.

import numpy as np

arr = np.array([10, np.nan, 30, np.nan, 50])

missing_count = np.isnan(arr).sum()
percentage = (missing_count / arr.size) * 100

print("Missing values:", missing_count)
print("Missing percentage:", percentage, "%")
```

```python
# Q55. Replace NaN values with the mean of the array.

import numpy as np

arr = np.array([10, 20, np.nan, 40, 50])

mean_value = np.nanmean(arr)
arr[np.isnan(arr)] = mean_value

print(arr)
```

```python
# Q56. Generate 100 random numbers and calculate their statistics.

import numpy as np

arr = np.random.randint(1, 101, size=100)

print("Mean:", np.mean(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
print("Sum:", np.sum(arr))
```

```python
# Q57. Create a 5x5 matrix and extract its diagonal elements.

import numpy as np

arr = np.arange(1, 26).reshape(5, 5)

print("Matrix:\n", arr)
print("Diagonal:", np.diag(arr))
```

```python
# Q58. Find duplicate values in an array.

import numpy as np

arr = np.array([10, 20, 30, 20, 40, 10, 50])

values, counts = np.unique(arr, return_counts=True)
duplicates = values[counts > 1]

print("Duplicates:", duplicates)
```

```python
# Q59. Calculate the Euclidean distance between two points.

import numpy as np

point1 = np.array([2, 3])
point2 = np.array([5, 7])

distance = np.linalg.norm(point1 - point2)

print("Euclidean distance:", distance)
```

```python
# Q60. Find the best sales day, total sales, and average sales.

import numpy as np

sales = np.array([1200, 1500, 900, 2000, 1800, 2500, 1700])

days = np.array([
    "Monday", "Tuesday", "Wednesday", "Thursday",
    "Friday", "Saturday", "Sunday"
])

best_day = np.argmax(sales)

print("Best sales day:", days[best_day])
print("Highest sales:", sales[best_day])
print("Total sales:", np.sum(sales))
print("Average sales:", np.mean(sales))
```
