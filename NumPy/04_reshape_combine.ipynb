```python
# Q31. Reshape an array of 12 elements into 3 rows and 4 columns.

import numpy as np

arr = np.arange(1, 13)

result = arr.reshape(3, 4)

print(result)
```

```python
# Q32. Convert a 2D array into a 1D array.

import numpy as np

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

result = arr.flatten()

print(result)
```

```python
# Q33. Transpose a matrix.

import numpy as np

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

print(arr.T)
```

```python
# Q34. Join two arrays using concatenate().

import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

result = np.concatenate((arr1, arr2))

print(result)
```

```python
# Q35. Stack two arrays vertically.

import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

result = np.vstack((arr1, arr2))

print(result)
```

```python
# Q36. Stack two arrays horizontally.

import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

result = np.hstack((arr1, arr2))

print(result)
```

```python
# Q37. Split an array into three equal parts.

import numpy as np

arr = np.arange(1, 10)

result = np.split(arr, 3)

print(result)
```

```python
# Q38. Create a 3×3 identity matrix.

import numpy as np

arr = np.eye(3)

print(arr)
```

```python
# Q39. Convert a 1D array into a column vector.

import numpy as np

arr = np.array([1, 2, 3, 4, 5])

result = arr.reshape(-1, 1)

print(result)
```

```python
# Q40. Flatten a 2D array using flatten() and ravel().
# Compare their behavior.

import numpy as np

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

result1 = arr.flatten()
result2 = arr.ravel()

print("Flatten:", result1)
print("Ravel:", result2)
```
